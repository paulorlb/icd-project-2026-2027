### Why this network analysis matters to the project

The project asks how formal research groups relate to the thematic structure of their scientific
production. Before comparing themes, we need to establish what the available records say about
collaboration. Coauthorship is one observable relation: it allows us to ask who is connected
through shared publications, how often pairs recur, whether the observed researchers form one or
several connected sets, and how much recorded coauthorship crosses formal group boundaries
(Newman, 2001). Two researchers may have the same publication count while collaborating with
very different sets of colleagues; a publication total alone cannot show that pattern.

This is an **exploratory** stage. We define the population and the relation, audit the records,
construct two complementary graphs, and describe their principal patterns. The outputs provide a
baseline for the later work on topics: they can identify comparisons worth making and help
formulate more precise questions about whether collaborative and thematic structures correspond.
The sequence is **authorship records -> relational model -> descriptive measures -> questions
for thematic analysis**. Each step changes what can be said from the data.

The present counts establish the structure of *observed* coauthorship within the supplied
Scopus-based universe. Their interpretation depends on the included researchers, the publication
period, the group assignments and the absence of external coauthors. The W07-W09 thematic
measures will add another empirical dimension; any claim that formal groups correspond to themes
will require evidence from that dimension and a justified comparison.


### What is a network? What is a graph?

A **network** describes entities through relations among them. In empirical research, the first
task is to say which entities are included, which relation is observed, and over what period. A
**graph** gives this choice a precise form: $G=(V,E)$, where $V$ is the set of **nodes** (also
called *vertices*) and $E$ is the set of **edges** (also called *links* or *ties*). An edge
states that a specified relation holds between two nodes. The definition of the edge belongs to
the research design; lines on a drawing have no meaning until that definition is given.

![Illustrative undirected graph with five nodes, four edges, a path, and an isolated node](./_assets/w06/graph_basics.svg)

*Figure 1. An illustrative graph, separate from the P1-P4 exercise. Node A has degree 2; A-C-D is a path; E is an isolate. The graph has two connected components. Four of the ten possible node pairs are linked, giving density 0.40.*

Consider publication P1 in the fictitious example below. A and B both sign it. In the authorship
graph, A and B are linked separately to P1; the graph does not yet contain an A-B edge. After
projection, A and B are joined because they share P1. These are two representations of the same
records, answering questions at different units. A node such as E can remain in the researcher
graph even if it has no observed researcher-to-researcher edge.

Nodes can also carry **attributes**, such as a researcher's formal group. An attribute describes
a node; an edge describes a relation. Thus belonging to PI2 is not itself a collaboration tie.
Conversely, the absence of an edge means that the defined relation was not recorded *within this
dataset and period*. It does not establish that the two researchers never interacted.


### Types of graphs: the choices that determine an interpretation

Graph types describe different properties of a representation. These properties can be combined:
a graph may, for example, be undirected, simple and weighted at the same time. The choices below
follow the observational question and remain fixed for the main analysis.

![Six schematic graph types: undirected, directed, weighted, multigraph, bipartite, and temporal](./_assets/w06/graph_types.svg)

*Figure 2. Schematic graph types. These properties describe different dimensions and can coexist in one graph; the table specifies the choices used in this analysis.*

| Property | General distinction | Choice in this notebook |
|---|---|---|
| Kinds of node | A **one-mode** graph has one class of node; a **two-mode** or **bipartite** graph has two classes with links across them | Authorship is researcher-publication; projected coauthorship is researcher-researcher |
| Direction | A **directed** edge has an origin and a destination; an **undirected** edge has no direction | Signing and joint authorship are represented by undirected edges |
| Multiplicity | A **simple** graph has at most one edge for each eligible pair; a **multigraph** can have parallel edges | Each researcher pair has one edge; repeated joint publications increase its weight |
| Weight | A **binary** edge records presence; a **weighted** edge also records a magnitude | Authorship links are binary; projected coauthorship weight counts distinct shared publications |
| Time | A **temporal** graph retains when ties occur; a **static** graph pools observations | The main graph pools 2020-2026 publications into one network |

A self-loop joins a node to itself. It has no coauthorship meaning here, so the diagonal of the
projected matrix is retained separately as publication counts and removed from the graph.
Researchers with no projected edge remain as nodes: dropping them would change the population
and every denominator based on it.


### The bipartite graph: researchers linked to publications

A graph is **bipartite** when its nodes can be divided into two disjoint sets and every edge
joins one set to the other. In this case the sets are researchers $R$ and publications $P$.
Formally, $V=R\cup P$; each authorship edge has one endpoint in $R$ and the other in $P$. There are no researcher-researcher or publication-publication
edges in this original graph. The term *two-mode network* refers to the same arrangement of two
kinds of node (Latapy, Magnien & Del Vecchio, 2008).

The path `A - P1 - B` passes through publication P1: it consists of the two authorship links
A-P1 and B-P1. The direct A-B coauthorship edge appears only when we **project** the graph onto
researchers. A publication signed by several observed researchers can create several projected
pairs; a publication with one observed researcher creates none. Yet the latter publication still
matters for publication counts and for questions asked at the publication level.

![Researcher-publication bipartite graph and its weighted researcher projection for a fictitious example](./_assets/w06/bipartite_projection.svg)

*Figure 3. The fictitious P1-P4 authorship graph and its researcher projection. A-B has weight 2 because the pair shares P1 and P2. E signs P4 alone within the observed set and therefore has no projected coauthorship edge.*

Keeping the bipartite graph as the starting point makes the derivation of every coauthorship tie
traceable to the underlying publications. It also preserves the event that brought a team
together, which a list of researcher pairs alone cannot fully reconstruct. Section 1 builds this
graph from a small fictitious table before the same construction is applied to GOVCOPP.


### How the same network can be represented

A graph can be stored and inspected in several forms. They encode related information, but they
serve different tasks. In this notebook we move between the forms rather than treating a drawing
as the data itself.

| Representation | What it records | Example in this notebook |
|---|---|---|
| **Edge list** | One row per observed link, with identifiers for its two ends | The authorship CSV contains publication-researcher links; the exported coauthorship CSV contains researcher pairs and weights |
| **Incidence matrix** | Researchers in rows, publications in columns, with 1 for an authorship link and 0 otherwise | $B$ retains the two-mode relation; row and column sums have different meanings |
| **Adjacency matrix** | The same kind of node in rows and columns; a cell records a link or its weight | $B B^\top$ counts shared publications; after removing its publication-count diagonal, the off-diagonal entries give weighted adjacency |
| **Graph object** | Node and edge sets, with any attributes and weights needed for calculation | `toy_bipartite` and `G` support counts, components and other measures |
| **Drawing** | A visual placement of the nodes and edges | The figures help inspect the result; their coordinates are chosen for display |

### From a research question to an observable network

What, precisely, is the relation we can observe? The project asks about the relation between
formal organisation and thematic production. This session establishes one part of that inquiry:
**recorded joint authorship**. A node denotes a researcher represented in the supplied tables;
an authorship link denotes a researcher signing a publication; a coauthorship edge is derived
when two represented researchers sign the same publication. These are operational definitions,
rather than direct observations of intellectual proximity, friendship, supervision, or the
division of work within a paper.

The network is therefore a model of a specified relation within a specified boundary. The
boundary determines which publications and researchers can enter the calculation. A path, an
isolate, or a dense set of ties must always be read relative to that boundary and to the period
covered by the records. The analysis can describe the pattern produced by these choices; the
later thematic analysis is needed to examine whether that pattern corresponds to similarities in
research topics. This distinction between an empirical indicator and the wider concept it may
inform guides the interpretation of every measure below (Newman, 2001).

**In symbols.** Write $R$ for the set of researchers and $P$ for the set of publications. The
bipartite network has the nodes $R \cup P$ (read: "R union P", both kinds together) and one link
for every researcher–publication pair in which the researcher signs the publication. The
**Researcher × Publication Incidence (Biadjacency) Matrix** $B$ stores the same links as numbers, one row per researcher and one column
per publication:

$$B_{rp} = \begin{cases} 1 & \text{if researcher } r \text{ signs publication } p \\ 0 & \text{otherwise} \end{cases}$$

Read: "B r p is one when r signs p". The number of nodes is $\lvert R \rvert + \lvert P \rvert$ (the
vertical bars mean "how many"), and the number of links is the number of ones in $B$. In GOVCOPP,
$R$ holds the pseudonymised researchers, $P$ the publications of the textual universe, and each
one in $B$ is one row of the authorship table.

### Why authorship has two modes

The primary observation is an **event of participation**: a researcher appears on a publication.
Researchers and publications are different kinds of entity, so the original graph is *two-mode*
or *bipartite*. Its links run between the two sets, never between two researchers or two
publications. This representation preserves which particular publication brought each set of
researchers together. It also keeps a publication with only one observed GOVCOPP researcher,
although that publication will create no coauthor pair inside the observed researcher set.

The row sum of $B$ counts the publications attributed to a researcher in this table. The column
sum counts the GOVCOPP researchers represented on a publication. The latter is **not necessarily
the total number of authors on the paper**, because external coauthors are outside the
researcher set. These two degrees answer different questions and have different units. Retaining
the two-mode data allows us to return from a pattern of researcher pairs to the publications
from which those pairs were derived (Latapy, Magnien & Del Vecchio, 2008).


The same relation as a graph with two kinds of node. The prefixes `R:` and `P:` stop a
researcher and a publication from ever sharing a name. The **incidence matrix** is the same
information as a table: researchers in the rows, publications in the columns, 1 if the
researcher signs the publication and 0 otherwise.

````python
toy_bipartite = nx.Graph()
toy_bipartite.add_nodes_from(("R:" + r for r in sorted(TOY_GROUP)), kind="researcher")
toy_bipartite.add_nodes_from(("P:" + p for p in sorted(toy["publication_id"].unique())), kind="publication")
toy_bipartite.add_edges_from(("R:" + r, "P:" + p) for p, r in toy.itertuples(index=False))

toy_incidence = pd.crosstab(toy["researcher_id"], toy["publication_id"])
print(toy_bipartite.number_of_nodes(), "nodes,", toy_bipartite.number_of_edges(), "authorship links")
toy_incidence
````

## 2 From authorship to coauthorship: the projection

Two researchers are **coauthors** if they sign at least one publication together. The
**projection** keeps only researchers and joins two of them when they share a publication. The
**weight** of the edge is the number of distinct publications they share.

The same result with a matrix product. Multiplying the incidence matrix by its transpose,
`B @ B.T`, counts for every pair of researchers how many publications both sign. On the
diagonal, a researcher is paired with themself: that cell counts their publications. We keep
the diagonal as the number of publications and then set it to zero, because a researcher is not
their own coauthor.

**In symbols.** $B^\top$ ("B transposed") is $B$ with rows and columns swapped. The product

$$C = B\,B^\top, \qquad C_{ij} = \sum_{p \in P} B_{ip}\,B_{jp}$$

reads: "for researchers $i$ and $j$, go through every publication $p$ and add one when both sign
it". The product $B_{ip}\,B_{jp}$ is 1 only when both entries are 1, so the sum counts shared
publications. Off the diagonal ($i \neq j$), $C_{ij}$ is the edge weight $w_{ij}$. On the
diagonal, $C_{ii} = \sum_p B_{ip}$ is the number of publications of $i$: we keep it as its own
column and then set it to zero.

Seen from one publication: a publication signed by $n_p$ researchers adds one unit of weight to
each of its $\binom{n_p}{2} = n_p(n_p - 1)/2$ pairs (read: "$n_p$ choose 2", the number of
different pairs among $n_p$ people). P2 has $n_p = 3$ and adds 3 pairs; P4 has $n_p = 1$ and adds
none. You will use this rule again in section 6.

### What the projection preserves and what it loses

Projection changes the unit from researcher–publication participation to researcher pairs. It
preserves whether a pair has published together and, here, how many distinct publications the
pair shares. It does **not** preserve the event structure: the three edges among A, B and C
could arise from one paper signed by all three, from three separate two-author papers, or from a
combination of both. The projected edges alone cannot tell these situations apart. Consequently,
questions about the number or composition of multigroup publications must return to the
authorship links.

A paper with $n_p$ observed researchers creates $\binom{n_p}{2}$ pair contributions. Large teams
therefore exert a disproportionate influence on the total edge weight and may create a visually
dense cluster through a single publication. This notebook uses **full counting**: every pair on
that paper receives one unit. That is a defensible rule for the question “how many publications
did this pair share?”, but any interpretation of the aggregate network should acknowledge the
effect of team size. Fractional construction of bibliometric networks is a different possible
rule, with a different meaning for edge weights (Perianes-Rodríguez, Waltman & van Eck, 2016).

## 3 Degree, strength, components and isolates

- **Degree**: the number of *distinct* coauthors of a researcher.
- **Strength**: the sum of the weights of their edges (each shared publication counted once
  per coauthor).
- **Publications**: how many publications they sign. A different quantity again.
- A **component** is a set of researchers linked by some path of edges; an **isolate** is a
  researcher with no edge.
- **Density** = distinct edges / possible pairs of researchers, *counting the isolates* among
  the researchers.

**In symbols.** With $w_{ij}$ the weight of the edge between $i$ and $j$ (zero when there is no
edge) and $N$ the number of researchers:

$$k_i = \#\{\, j \neq i : w_{ij} > 0 \,\} \qquad s_i = \sum_{j \neq i} w_{ij} \qquad D = \frac{\lvert E \rvert}{\binom{N}{2}} = \frac{2\,\lvert E \rvert}{N(N-1)}$$

Read: the **degree** $k_i$ is "how many $j$ have a positive weight with $i$"; the **strength**
$s_i$ is "the sum of $i$'s weights"; the **density** $D$ is "distinct edges $\lvert E \rvert$ over
the number of possible pairs". Every edge weight is counted twice in the strengths, once at each
end, so $\sum_i s_i = 2 \sum_{i<j} w_{ij}$. That is a check you can run on any weighted network.

None of the three measures says anything about quality, and degree 0 only means "no coauthor
*among the researchers we observe*".

### Reading local and global measures

Degree, strength and publication count should be interpreted together because the same
publication can affect them differently. A paper with several observed researchers adds one
publication to each researcher, one partner to their degree for every *new* coauthor, and one
unit of strength for every coauthor on that paper. Repeated papers with an existing partner
increase strength without increasing degree. Thus $s_i \geq k_i$, while neither quantity is
interchangeable with the number of publications. In particular, strength can exceed publication
count when papers have several observed coauthors.

A component expresses **reachability by a chain of coauthorship edges**. If A published with B
and B with C, A and C can belong to the same component without ever sharing a paper. An isolate
has no observed coauthor in this defined universe; the term says nothing about collaborations
with authors outside it. Density divides the number of observed distinct pairs by every possible
pair among the included researchers. Its value changes when the network boundary changes, even
if the observed collaborations do not. These are descriptions of the constructed graph, rather
than measures of scientific merit or evidence that one researcher caused another to collaborate
(Newman, 2001).

## 6 The real coauthorship projection — checkpoint C3 (minutes 120–145)

Exactly the steps of section 2: `B @ B.T`, keep the diagonal as publications per researcher,
remove it, and build a simple, undirected, weighted network with **all** researchers.

````python
M = (B @ B.T).tocsr()
n_publications_per_researcher = M.diagonal().astype("int64")
M.setdiag(0)
M.eliminate_zeros()                                              # drop the zeros left on the diagonal
upper = sp.triu(M, k=1).tocoo()                                  # each unordered pair once
edges = pd.DataFrame({
    "researcher_a": [researchers[i] for i in upper.row],
    "researcher_b": [researchers[j] for j in upper.col],
    "weight": upper.data.astype("int64"),
}).sort_values(["researcher_a", "researcher_b"], ignore_index=True)

G = nx.Graph()
G.add_nodes_from(researchers)                                    # nodes first: isolates stay
G.add_weighted_edges_from(edges.itertuples(index=False))
print(G.number_of_nodes(), "researchers,", G.number_of_edges(), "edges, total weight", edges["weight"].sum())
````

## 8 Coauthorship and the formal groups — checkpoint C4 (minutes 145–170)

**What the literature says — the unit.** The unit of analysis is a choice: it should follow
from the question being asked, from the method, and from the form in which the data arrive
(Grimmer, Roberts & Stewart (2022), *Text as Data*, “5.2 Choose the Unit of Analysis”). Below, the same source records give two shares because they are read in two
representations, at two units: the projection (researcher pairs) and the authorship links
(publications).

Two different questions, with **different units and denominators**:

- *Edges* (from the projection): of all the edge weight between researchers whose group is valid,
  how much joins researchers of **different** groups? Unit: researcher pair × shared publication.
- *Publications* (from the authorship links): of all publications whose researchers all have a
  valid group, how many involve **two or more** groups? Unit: publication. The projection alone
  cannot answer this: it no longer knows which edges came from the same publication.

In the fictitious example the first share is 1/5 (only CD crosses G1–G2) and the second is 1/4
(only P3 involves both groups). P4 belongs to G2 alone but contains no G2 pair: a publication
exclusive to a group is not coauthorship inside the group.

**In symbols.** Write $g_i$ for the formal group of researcher $i$, $G_p$ for the set of formal
groups present in publication $p$, and $\mathbf{1}[\ldots]$ for "1 when the condition in brackets
holds, 0 otherwise". The two shares are:

$$\text{edge share} = \frac{\sum_{i<j} w_{ij}\,\mathbf{1}[g_i \neq g_j]}{\sum_{i<j} w_{ij}} \qquad\qquad \text{publication share} = \frac{\#\{\, p : \lvert G_p \rvert \geq 2 \,\}}{\lvert P \rvert}$$

Read the first as "weight between different groups over all weight", with sums only over pairs
whose two researchers have a valid group. Read the second as "publications with two or more
groups over all publications" (the denominator only counts publications whose researchers all
have a valid group). The numerators differ, the denominators differ, and so do the units:
researcher pair × shared publication against publication. That is why the two numbers are not
two estimates of the same thing, and why your sentence must name the unit.

### Formal groups, units and interpretation

The formal groups are **node attributes** imported from the data, while coauthorship edges are
relations constructed from shared publications. Counting an edge as within-group or
between-group compares a relation with an attribute. It does not test whether the groups are
network communities, whether their members study the same subjects, or whether membership
produced the collaboration. A multigroup publication is likewise a property of one publication,
even if it creates several between-group researcher pairs.

The two shares answer different descriptive questions. The edge share gives more influence to
publications that generate many researcher pairs and to pairs that publish together repeatedly.
The publication share gives one observation to each eligible publication, regardless of how many
pairs it creates. Both also depend on the sizes and publication volumes of the formal groups and
on who is included in the dataset. A value above or below one half has no intrinsic
interpretation as “high” or “low” integration: such a judgment would require an explicit
comparison or model with a justified baseline. Here, the valid conclusion is the observed
proportion under the stated construction and denominator.


````python
edge_groups = edges.assign(group_a=edges["researcher_a"].map(affiliation["formal_group"]),
                           group_b=edges["researcher_b"].map(affiliation["formal_group"]))
edge_groups["classifiable"] = edge_groups["group_a"].notna() & edge_groups["group_b"].notna()
edge_groups["between_groups"] = edge_groups["classifiable"] & (edge_groups["group_a"] != edge_groups["group_b"])
unclassifiable_weight = int(edge_groups.loc[~edge_groups["classifiable"], "weight"].sum())

pair = edge_groups[edge_groups["classifiable"]].assign(
    pair=lambda t: [" – ".join(sorted(x)) for x in zip(t["group_a"], t["group_b"])])
group_pairs = pair.groupby("pair").agg(researcher_pairs=("weight", "size"), sum_of_weights=("weight", "sum"))
print("Unit: pairs of researchers with at least one shared publication; weight = shared publications.")
print("Edge weight left out (unresolved affiliation):", unclassifiable_weight)
group_pairs
````

**Counting publications per group.** If each multigroup publication counts once *for every*
group it involves, the group totals add up to more than 3343. The **fractional publication
count** gives each publication a total of one, split equally among its groups, so the group
totals add back to the number of publications. (W08 uses the same rule for group topic profiles.)

**In symbols.** For a group $g$:

$$\text{whole: } n_g = \sum_{p \in P} \mathbf{1}[g \in G_p] \qquad\qquad \text{fractional: } f_g = \sum_{p \in P} \frac{\mathbf{1}[g \in G_p]}{\lvert G_p \rvert}$$

A publication with two groups adds 1 to each group's whole count, but only $\tfrac{1}{2}$ to each
fractional count. Summing over the groups, every publication contributes exactly one to the
fractional total, so $\sum_g f_g = \lvert P \rvert$, while $\sum_g n_g = \sum_p \lvert G_p \rvert$
exceeds $\lvert P \rvert$ whenever some publication has more than one group. The fractional count
is a way of *attributing* production to groups; it is not a sampling weight.

### Two distinct meanings of fractional counting

The fractional count above attributes **one publication** across the formal groups represented
on it. If a publication involves three groups, each receives one third in the group publication
table. This prevents the same publication from adding three units to the sum of group outputs.
It does not change the coauthorship network.

Fractional **network construction** would make a separate decision about how much weight a
publication contributes to each researcher pair, usually to prevent large teams from dominating
aggregate link weights. The present edge weights remain full counts of shared publications;
weighted edge totals and the between-group edge share use those weights. The group publication
fractions cannot be substituted into the edge-share formula: the objects being counted are
different. State which object is being divided before calling a result “fractional”
(Perianes-Rodríguez, Waltman & van Eck, 2016).


````python
group_incidence = link_groups.drop_duplicates(["publication_id", "group"]).dropna(subset=["group"])
group_incidence = group_incidence.assign(share=1 / group_incidence["publication_id"].map(n_groups))
by_group = group_incidence.groupby("group").agg(publications_whole=("publication_id", "size"),
                                                publications_fractional=("share", "sum"))
by_group
````

### Interpreting the threshold contrast

The threshold asks what happens to the description **if only pairs with at least two shared

closely in every substantive sense. A lost path can split a component even when all the original
publications remain in the data. The result is therefore sensitivity to an analytical rule, not
evidence that the underlying researchers changed their behaviour.

Because the same 187 researchers remain in both scenarios, changes in components and isolates
can be compared directly. The edge and weight denominators, however, change with the threshold.
